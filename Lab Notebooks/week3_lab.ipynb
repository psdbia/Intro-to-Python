## Part 1 — Conditional Statements

A **conditional** runs a block of code only when a condition is true.

```python
if condition:
    # runs when condition is True
elif other_condition:
    # runs when other_condition is True
else:
    # runs when no condition above is True
```

Indentation matters — Python uses it to define what belongs inside each block.

````python
# Classify a single month's sales performance
monthly_sales = 17200
target        = 15000

if monthly_sales >= target * 1.10:
    print("Exceeded target by 10% or more")
elif monthly_sales >= target:
    print("Met target")
else:
    print("Below target")
````

### Comparison Operators

| Operator | Meaning | Example |
|:---------|:--------|:--------|
| `==` | Equal to | `sales == target` |
| `!=` | Not equal to | `region != "West"` |
| `>` | Greater than | `sales > 10000` |
| `<` | Less than | `sales < 10000` |
| `>=` | Greater than or equal | `rating >= 4` |
| `<=` | Less than or equal | `discount <= 0.20` |

### Logical Operators

Combine conditions using `and`, `or`, and `not`.

```python
if sales > 10000 and region == "Northeast":
    print("High-performing Northeast sale")
```

## Part 2 — For Loops

A **for loop** repeats a block of code once for each item in a collection.

```python
for item in collection:
    # code that runs for each item
```

````python
import numpy as np

monthly_sales = [14200, 13850, 16400, 15900, 17200, 18100]
months        = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
target        = 15000

# Print each month's sales and whether it met the target
for i in range(len(monthly_sales)):
    status = "✓" if monthly_sales[i] >= target else "✗"
    print(f"{months[i]}: ${monthly_sales[i]:,}  {status}")
````

### Looping with `enumerate()`

`enumerate()` gives you both the index and the value in each iteration — cleaner than using `range(len(...))`.

````python
for i, sales in enumerate(monthly_sales):
    print(f"{months[i]}: ${sales:,}")
````

### Accumulating Results in a Loop

A common pattern: start with an empty list, then append results inside the loop.

````python
# Collect only the months that exceeded the target
above_target = []

for i, sales in enumerate(monthly_sales):
    if sales >= target:
        above_target.append(months[i])

print("Months above target:", above_target)
````

## Part 3 — List Comprehensions

A **list comprehension** creates a new list by applying an expression to each item in a collection. It is a compact alternative to a loop.

```python
# Loop version
result = []
for item in collection:
    result.append(expression)

# Comprehension version — same result, one line
result = [expression for item in collection]
```

You can add a filter condition:

```python
result = [expression for item in collection if condition]
```

````python
# Apply a 5% growth projection to every month
projected = [sales * 1.05 for sales in monthly_sales]
print("Projected:", projected)

# Filter to only months above the target
above_target = [sales for sales in monthly_sales if sales >= target]
print("Above target:", above_target)

# Filter and transform — get projected values only for above-target months
projected_above = [sales * 1.05 for sales in monthly_sales if sales >= target]
print("Projected (above target only):", projected_above)
````

## Part 4 — Frequency Distributions

### 📊 Stats Connection — Frequency Distributions

A **frequency distribution** shows how often each value (or range of values) appears in a dataset. It is one of the most fundamental tools for understanding the shape of your data before doing any deeper analysis.

| Type | Used For | Example |
|:-----|:---------|:--------|
| **Frequency** | Count of occurrences | 42 transactions in the Northeast region |
| **Relative frequency** | Proportion of total | Northeast = 35% of all transactions |
| **Cumulative frequency** | Running total | First 3 categories account for 80% of sales |

Python's `dict` (dictionary) is the natural tool for building frequency tables.

````python
# Sales transactions by region
regions = [
    "Northeast", "West", "Northeast", "South", "West",
    "Northeast", "Midwest", "West", "South", "Northeast",
    "West", "Northeast", "South", "Midwest", "West",
    "Northeast", "West", "South", "Northeast", "Midwest"
]

# Build a frequency table using a dictionary
freq = {}
for region in regions:
    if region in freq:
        freq[region] += 1
    else:
        freq[region] = 1

print("Frequency table:")
for region, count in freq.items():
    print(f"  {region}: {count}")
````

````python
# Relative frequency — proportion of total
total = len(regions)

print("Relative frequency table:")
for region, count in freq.items():
    relative = (count / total) * 100
    print(f"  {region}: {count} ({relative:.1f}%)")
````

### Using `Counter` — A Shortcut

Python's `collections.Counter` builds a frequency table in one line.

````python
from collections import Counter

freq_counter = Counter(regions)
print(freq_counter)

# Most common regions
print("\nTop 3 regions:")
for region, count in freq_counter.most_common(3):
    print(f"  {region}: {count}")
````

## Part 5 — Outlier Detection

### 📊 Stats Connection — Outliers

An **outlier** is a value that falls unusually far from the rest of the data. Outliers can signal:
- A genuine exceptional event (a record sale, a data entry error)
- A problem in the data collection process
- A meaningful pattern worth investigating

A simple threshold rule: flag any value more than **1.5 times the interquartile range (IQR)** beyond the quartiles. We will cover IQR in depth in Week 4 — for now, we will use a simpler approach: flag values more than **2 standard deviations from the mean**.

$$\text{outlier if: } |x - \bar{x}| > 2\sigma$$

````python
# Daily transaction values for one month
transactions = np.array([
    420, 380, 510, 460, 390, 475, 502, 418, 395, 440,
    465, 388, 520, 410, 3800, 455, 430, 498, 375, 485,
    412, 468, 390, 445, 502, 415, 480, 395, 425, 460
])

mean = np.mean(transactions)
std  = np.std(transactions)

print(f"Mean:  ${mean:.2f}")
print(f"Std:   ${std:.2f}")
print(f"Upper threshold (mean + 2*std): ${mean + 2*std:.2f}")
print(f"Lower threshold (mean - 2*std): ${mean - 2*std:.2f}")
````

````python
# Flag outliers using a list comprehension
outliers = [x for x in transactions if abs(x - mean) > 2 * std]
print("Outliers detected:", outliers)
````
