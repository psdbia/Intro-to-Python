## Part 1 — Python Lists

Last week we stored one value per variable. In practice, data comes in collections — a column of sales figures, a list of product names, a set of monthly totals.

A **list** stores multiple values in a single variable, in order.

```python
my_list = [value1, value2, value3]
```

Lists can hold any data type — integers, floats, strings, or a mix.

````python
# Monthly sales figures for one product (January through June)
monthly_sales = [14200, 13850, 16400, 15900, 17200, 18100]

# Product names in a catalogue
products = ["Laptop Stand", "Monitor Arm", "Keyboard Tray", "Cable Manager", "Desk Lamp"]

print(monthly_sales)
print(products)
````

### List Properties

Use `len()` to get the number of items, and `sum()` to total a numeric list.

````python
print(len(monthly_sales))    # number of months
print(sum(monthly_sales))    # total sales across all months
````

---
## Part 2 — Indexing and Slicing

Each item in a list has a position called an **index**. Python indexing starts at **0**, not 1.

```
monthly_sales = [14200, 13850, 16400, 15900, 17200, 18100]
index:              0      1      2      3      4      5
```

Use square brackets to access an item by index.

````python
print(monthly_sales[0])     # January  — first item
print(monthly_sales[5])     # June     — last item
print(monthly_sales[-1])    # June     — negative index counts from the end
print(monthly_sales[-2])    # May
````

### Slicing

A **slice** extracts a portion of a list using `[start:stop]`. The stop index is **not included**.

```python
my_list[start:stop]    # items from start up to (not including) stop
```

````python
# Q1 = January, February, March (indices 0, 1, 2)
q1_sales = monthly_sales[0:3]
print("Q1:", q1_sales)

# Q2 = April, May, June (indices 3, 4, 5)
q2_sales = monthly_sales[3:6]
print("Q2:", q2_sales)

# Omitting start or stop defaults to the beginning or end
print(monthly_sales[:3])    # first three months
print(monthly_sales[3:])    # last three months
````

## Part 3 — Introduction to NumPy

Python lists are flexible but slow for large numerical datasets. **NumPy** (Numerical Python) provides the `array` — a faster, more powerful structure designed for numerical data.

NumPy arrays are the foundation of most data science work in Python. They also come with built-in statistical functions.

Import NumPy with the standard alias `np`.

````python
import numpy as np

# Convert the monthly sales list to a NumPy array
sales_array = np.array(monthly_sales)

print(sales_array)
print(type(sales_array))
````

### Array Arithmetic

One major advantage of NumPy arrays: arithmetic operates on **every element at once**, without needing a loop.

````python
# Apply a 10% increase to every month's sales
adjusted_sales = sales_array * 1.10
print("Adjusted sales:", adjusted_sales)

# Subtract a fixed monthly overhead from each month
overhead = 2000
net_sales = sales_array - overhead
print("Net sales:", net_sales)
````

## Part 4 — Samples and Populations

### 📊 Stats Connection — Samples vs Populations

In statistics, a **population** is the complete set of all cases you are interested in. A **sample** is a subset of the population that you actually have data for.

| | Population | Sample |
|:--|:-----------|:-------|
| **Definition** | All cases of interest | A subset of the population |
| **Size notation** | $N$ | $n$ |
| **Mean notation** | $\mu$ (mu) | $\bar{x}$ (x-bar) |
| **Example** | All transactions in company history | Last 6 months of transactions |

In practice you almost always work with samples. This matters because some statistical formulas differ slightly depending on whether you have a sample or the full population — you will see this again in Week 4 when we cover variance and standard deviation.

The six months of sales data in this lab is a **sample** — it represents part of a longer sales history.

````python
# Our sample: 6 months of sales data
# n = 6 (sample size)

n = len(sales_array)
print("Sample size (n):", n)
````

## Part 5 — Measures of Central Tendency

A measure of central tendency summarises a dataset with a single value that represents the "centre" of the data. There are three you need to know.

### 📊 Stats Connection — Mean, Median, Mode

| Measure | Definition | Best Used When |
|:--------|:-----------|:---------------|
| **Mean** | Sum divided by count | Data is symmetric, no extreme outliers |
| **Median** | Middle value when sorted | Data is skewed or has outliers |
| **Mode** | Most frequently occurring value | Categorical data, or finding the most common value |

---

### The Mean

$$\bar{x} = \frac{\sum x}{n}$$

````python
# Mean — manually and with NumPy
mean_manual = sum(monthly_sales) / len(monthly_sales)
mean_numpy  = np.mean(sales_array)

print(f"Mean (manual): {mean_manual}")
print(f"Mean (NumPy):  {mean_numpy}")
````

### The Median

The median is the middle value of a sorted dataset.

- If $n$ is odd: the median is the middle value
- If $n$ is even: the median is the average of the two middle values

$$\text{Median position} = \frac{n+1}{2}$$

````python
# Median
median_sales = np.median(sales_array)
print(f"Median: {median_sales}")

# Sort the array to see what NumPy is doing
sorted_sales = np.sort(sales_array)
print(f"Sorted: {sorted_sales}")
print(f"n = {len(sorted_sales)} — even number, so median = average of positions 3 and 4")
print(f"Manual check: ({sorted_sales[2]} + {sorted_sales[3]}) / 2 = {(sorted_sales[2] + sorted_sales[3]) / 2}")
````

### The Mode

The mode is the value that appears most often. It is the only measure of central tendency that works with **nominal** data.

````python
from scipy import stats

# Sales rep performance ratings for a team of 10
ratings = np.array([3, 4, 4, 5, 3, 4, 2, 5, 4, 3])

mode_result = stats.mode(ratings, keepdims=True)
print(f"Mode: {mode_result.mode[0]}")
print(f"Appears: {mode_result.count[0]} times")
````

### When Mean and Median Differ — Outliers

The mean is sensitive to extreme values (outliers). The median is not. Watch what happens when one month has an unusually high sale.

````python
# Normal monthly sales
normal_sales = np.array([14200, 13850, 16400, 15900, 17200, 18100])

# Same data but one month had an exceptional bulk order
outlier_sales = np.array([14200, 13850, 16400, 15900, 17200, 95000])

print("--- Without outlier ---")
print(f"Mean:   {np.mean(normal_sales):.2f}")
print(f"Median: {np.median(normal_sales):.2f}")

print("\n--- With outlier ---")
print(f"Mean:   {np.mean(outlier_sales):.2f}")
print(f"Median: {np.median(outlier_sales):.2f}")
````

> **Notice:** The outlier pulls the mean up significantly, but the median barely moves. When reporting on data with outliers — like executive salaries, property prices, or one-off bulk orders — the **median is often a more honest summary** than the mean.
