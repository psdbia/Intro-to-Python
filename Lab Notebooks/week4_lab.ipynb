## Part 1 — Writing Functions

A **function** is a reusable block of code. Instead of repeating the same logic in multiple places, you write it once and call it by name.

```python
def function_name(parameter1, parameter2):
    # code
    return result
```

- `def` declares the function
- **Parameters** are inputs the function receives
- `return` sends a value back to the caller

````python
import numpy as np

# A function that calculates percentage change between two values
def pct_change(old, new):
    change = ((new - old) / old) * 100
    return change

print(pct_change(48000, 54600))    # 13.75
print(pct_change(15000, 13500))    # -10.0
````

### Default Parameters

Parameters can have **default values**. If the caller does not provide that argument, the default is used.

````python
# Apply a growth rate to a revenue figure — default rate is 5%
def project_revenue(revenue, growth_rate=0.05):
    return revenue * (1 + growth_rate)

print(project_revenue(100000))          # 105000.0 — uses default 5%
print(project_revenue(100000, 0.12))    # 112000.0 — uses 12%
````

### Returning Multiple Values

A function can return more than one value as a tuple.

````python
def summary_stats(data):
    return np.mean(data), np.median(data), np.std(data)

sales = [14200, 13850, 16400, 15900, 17200, 18100]
mean, median, std = summary_stats(sales)

print(f"Mean:   {mean:.2f}")
print(f"Median: {median:.2f}")
print(f"Std:    {std:.2f}")
````

## Part 2 — Error Handling

Functions can fail if given unexpected input. `try / except` lets you handle errors gracefully instead of crashing.

```python
try:
    # code that might fail
except ErrorType:
    # what to do if it fails
```

````python
# Division by zero is a common real-world issue in business calculations
def safe_pct_change(old, new):
    try:
        change = ((new - old) / old) * 100
        return round(change, 2)
    except ZeroDivisionError:
        return "Error: base value cannot be zero"

print(safe_pct_change(50000, 55000))    # 10.0
print(safe_pct_change(0, 55000))        # Error message
````

````python
# Handle wrong data types gracefully
def safe_mean(data):
    try:
        return sum(data) / len(data)
    except TypeError:
        return "Error: data must be a list of numbers"
    except ZeroDivisionError:
        return "Error: list is empty"

print(safe_mean([100, 200, 300]))           # 200.0
print(safe_mean(["a", "b", "c"]))           # TypeError
print(safe_mean([]))                         # ZeroDivisionError
````

## Part 4 — Variance and Standard Deviation

### 📊 Stats Connection — Spread

The mean tells you where the centre of the data is. **Variance** and **standard deviation** tell you how spread out the data is around that centre.

Two sales teams can have the same mean revenue but very different consistency — one team might be reliably hitting targets while another swings wildly. Spread measures capture that difference.

### Variance

Variance is the average squared deviation from the mean.

**Population variance:**
$$\sigma^2 = \frac{\sum (x - \mu)^2}{N}$$

**Sample variance** (uses $n-1$ to correct for the fact that a sample underestimates spread):
$$s^2 = \frac{\sum (x - \bar{x})^2}{n-1}$$

### Standard Deviation

Standard deviation is the square root of variance. It brings the units back to the original scale (dollars, not dollars squared).

$$s = \sqrt{s^2}$$

````python
# Two sales reps — same mean, different consistency
rep_a = np.array([15200, 15800, 14900, 16100, 15400, 15600])
rep_b = np.array([9800,  21000, 12500, 18900, 11200, 19600])

print(f"Rep A — Mean: {np.mean(rep_a):.1f}  Std: {np.std(rep_a, ddof=1):.1f}")
print(f"Rep B — Mean: {np.mean(rep_b):.1f}  Std: {np.std(rep_b, ddof=1):.1f}")
````

> **Note:** `ddof=1` tells NumPy to use sample standard deviation ($n-1$). The default (`ddof=0`) uses population standard deviation ($N$). Use `ddof=1` whenever you are working with a sample — which is almost always.

````python
# Calculate variance and std manually to see the steps
data = rep_a
mean = np.mean(data)
n    = len(data)

deviations        = data - mean
squared_devs      = deviations ** 2
sample_variance   = np.sum(squared_devs) / (n - 1)
sample_std        = np.sqrt(sample_variance)

print(f"Mean:             {mean:.2f}")
print(f"Deviations:       {deviations}")
print(f"Squared devs:     {squared_devs}")
print(f"Sample variance:  {sample_variance:.2f}")
print(f"Sample std:       {sample_std:.2f}")
````

## Part 5 — Z-Scores

### 📊 Stats Connection — Standardisation

A **z-score** measures how many standard deviations a value is from the mean.

$$z = \frac{x - \bar{x}}{s}$$

Z-scores are useful for:
- Comparing values across different scales (e.g., comparing rep performance across regions with different typical sales volumes)
- Identifying outliers — values with $|z| > 2$ or $|z| > 3$ are commonly flagged
- Standardising data before further analysis

A z-score of `+1.5` means the value is 1.5 standard deviations above the mean. A z-score of `-2.1` means it is 2.1 standard deviations below.

````python
# Calculate z-scores for regional sales figures
regional_sales = np.array([42100, 38900, 45300, 31800, 39200, 48600, 36500, 44100])

mean   = np.mean(regional_sales)
std    = np.std(regional_sales, ddof=1)
z_scores = (regional_sales - mean) / std

print(f"Mean: {mean:.2f}  Std: {std:.2f}")
print()
for sales, z in zip(regional_sales, z_scores):
    flag = " ← flagged" if abs(z) > 2 else ""
    print(f"${sales:,}  z = {z:+.2f}{flag}")
````

## Part 6 — IQR and Robust Outlier Detection

### 📊 Stats Connection — The Interquartile Range

The **interquartile range (IQR)** is the range of the middle 50% of the data — from the 25th percentile (Q1) to the 75th percentile (Q3).

$$IQR = Q3 - Q1$$

The **Tukey fence** rule flags outliers as values below $Q1 - 1.5 \times IQR$ or above $Q3 + 1.5 \times IQR$.

Unlike the 2 standard deviation rule, the IQR method is not affected by outliers themselves — making it more reliable when the data is skewed or already contains extreme values.

````python
# IQR outlier detection
sales_data = np.array([4200, 3850, 4600, 4100, 3920, 12400, 4350, 3780, 4500, 4050, 3900, 4250])

q1  = np.percentile(sales_data, 25)
q3  = np.percentile(sales_data, 75)
iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

print(f"Q1:           {q1}")
print(f"Q3:           {q3}")
print(f"IQR:          {iqr}")
print(f"Lower fence:  {lower_fence}")
print(f"Upper fence:  {upper_fence}")

outliers = sales_data[(sales_data < lower_fence) | (sales_data > upper_fence)]
print(f"Outliers:     {outliers}")
````
