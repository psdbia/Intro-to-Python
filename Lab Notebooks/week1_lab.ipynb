### 📊 Stats Connection — Measurement Scales

In statistics, every variable has a **measurement scale** that determines what operations make sense on it.

| Scale | Description | Python Type | Example |
|:------|:------------|:------------|:--------|
| **Nominal** | Categories with no order | `str` | Product name, region |
| **Ordinal** | Categories with order, but gaps aren't equal | `str` or `int` | Customer rating (1–5 stars) |
| **Interval** | Ordered, equal gaps, no true zero | `float` or `int` | Year, temperature (°C) |
| **Ratio** | Ordered, equal gaps, true zero exists | `float` or `int` | Revenue, units sold |

> **Why it matters:** You can calculate an average revenue (ratio), but averaging postal codes (nominal) is meaningless — even though both are stored as numbers.

## Part 3 — Arithmetic Operations

Python supports the arithmetic you'd expect, plus a couple of extras:

| Operator | Operation | Example | Result |
|:---------|:----------|:--------|:-------|
| `+` | Addition | `10 + 3` | `13` |
| `-` | Subtraction | `10 - 3` | `7` |
| `*` | Multiplication | `10 * 3` | `30` |
| `/` | Division | `10 / 3` | `3.333...` |
| `//` | Floor division | `10 // 3` | `3` |
| `%` | Remainder (modulo) | `10 % 3` | `1` |
| `**` | Exponentiation | `10 ** 3` | `1000` |

````python
# Sales figures for a small product line

q1_revenue = 12400.00
q2_revenue = 15800.00
q3_revenue = 13950.00
q4_revenue = 19200.00

# Total annual revenue
annual_revenue = q1_revenue + q2_revenue + q3_revenue + q4_revenue
print("Annual Revenue:", annual_revenue)

# Simple average (mean) — sum divided by number of quarters
average_quarterly_revenue = annual_revenue / 4
print("Average Quarterly Revenue:", average_quarterly_revenue)
````

---
### 📊 Stats Connection — The Mean

The **mean** (average) is the sum of all values divided by the count of values.

$$\bar{x} = \frac{\sum x}{n}$$

You just calculated one. The mean is only meaningful for **interval** and **ratio** scale variables.

## Part 4 — String Operations

Strings store text data — product names, regions, categories. A few operations you'll use often:

````python
product  = "laptop stand"
region   = "Northeast"

# Capitalisation
print(product.upper())       # LAPTOP STAND
print(product.title())       # Laptop Stand

# String length
print(len(product))          # 12

# Concatenation — joining strings
label = product.title() + " — " + region
print(label)                 # Laptop Stand — Northeast

# f-strings — cleaner way to embed variables in text
units = 47
print(f"{product.title()} sold {units} units in {region}.")
````
